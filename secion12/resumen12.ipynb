Investigación: Árboles de Decisión y Modelos de Caja Blanca
1. ¿Qué son los Árboles de Decisión?
Un árbol de decisión es un algoritmo de aprendizaje supervisado que se utiliza principalmente para problemas de clasificación y regresión. Su estructura imita la de un árbol real:

Nodo raíz y nodos de decisión: Representan preguntas o pruebas sobre una característica específica de los datos (por ejemplo, si la longitud del pétalo es menor o igual a un valor determinado).

Ramas (Rutas de decisión): Indican el resultado de la prueba (Verdadero/True o Falso/False), dirigiendo el flujo hacia el siguiente nivel.

Nodos hoja (Leaf nodes): Representan la decisión final o la clase asignada a la predicción.

2. Modelos de Caja Blanca
Como se observa en la presentación, los árboles de decisión forman parte de los modelos de caja blanca. Esto significa que:

Tienen reglas de clasificación transparentes.

Son fáciles de interpretar paso a paso, permitiendo comprender exactamente por qué el modelo tomó una decisión específica.

A diferencia de los modelos de "caja negra" (como las redes neuronales profundas), cualquier persona puede visualizar y seguir la lógica interna del modelo de forma gráfica.

Investigación Profunda: Árboles de Decisión (Modelos de Caja Blanca)
Módulo 01: Entrenando Árboles (Dentro de la caja blanca)
El entrenamiento de un árbol de decisión consiste en encontrar la estructura óptima (las preguntas correctas y el orden en el que se deben hacer) para separar los datos de la manera más pura posible.

El proceso iterativo: El algoritmo comienza en el nodo raíz con todo el conjunto de datos. Evalúa todas las características disponibles (por ejemplo, longitud y ancho de pétalos o sépalos) y calcula qué variable y qué valor de corte divide a los datos en dos grupos de forma que queden lo más homogéneos posible.

Algoritmo recursivo (Divide y vencerás): Una vez hecha la primera división, el proceso se repite en cada subgrupo resultante (nodos hijos) de manera independiente. Este proceso continúa hasta que se cumple una condición de parada (por ejemplo, que todos los elementos de un nodo pertenezcan a una sola clase, o que se alcance una profundidad máxima predefinida).

Módulo 02: Predicciones y ProbabilidadesUna vez que el árbol está entrenado, hacer una predicción para un nuevo dato es un proceso sumamente intuitivo y rápido:Recorrido paso a paso: El nuevo dato ingresa por el nodo raíz y responde a la pregunta planteada (ej. ¿Es la longitud del pétalo $\le 2.45\text{ cm}$?). Dependiendo de si la respuesta es Verdadero o Falso, el dato baja por la rama correspondiente hasta llegar a un nodo intermedio o a una hoja.Cálculo de probabilidades: En el nodo hoja final, el modelo no solo dice una categoría, sino que calcula una probabilidad basada en las muestras de entrenamiento que cayeron en esa misma hoja. Por ejemplo, si una hoja contiene 45 muestras de la clase Versicolor y 5 muestras de la clase Virginica, el modelo predecirá Versicolor con una probabilidad del $90\%$ ($45/50$) y Virginica con un $10\%$.

Módulo 03: CART e Impureza (Medidas de división)Para decidir cuál es el mejor corte en cada paso, el algoritmo necesita una métrica matemática que mida el "desorden" o la mezcla de clases en un conjunto de datos. Aquí es donde entra en juego el algoritmo CART (Classification and Regression Trees) y el Índice de Gini.¿Qué es la Impureza de Gini?Mide la probabilidad de que un elemento elegido al azar sea mal clasificado si se le asignara una etiqueta aleatoria basada en la distribución de las clases del nodo.Un valor de Gini = 0 significa que el nodo es puro (todos los elementos pertenecen a una sola clase).Un valor de Gini alto indica que las clases están completamente mezcladas.La fórmula de Gini:$$\text{Gini} = 1 - \sum_{i=1}^{c} (p_i)^2$$(donde $p_i$ es la proporción de elementos de la clase $i$ en el nodo).Ganancia de información / Reducción de impureza: CART evalúa todos los cortes posibles y elige aquel que minimice la impureza combinada de los nodos hijos, maximizando así la ganancia de separación.

Módulo 04: Regularización (Controlando el Overfitting)
Los árboles de decisión tienen una tendencia natural a "sobreajustarse" (overfitting) si se les permite crecer sin restricciones. Un árbol muy profundo memorizará los datos de entrenamiento (incluyendo el ruido), pero funcionará mal con datos nuevos. Para evitar esto, se aplican técnicas de regularización mediante hiperparámetros:

max_depth (Profundidad máxima): Limita cuántos niveles puede tener el árbol como máximo.

min_samples_split: Define el número mínimo de muestras que debe tener un nodo para que se le permita intentar una nueva división.

min_samples_leaf: Establece el número mínimo de muestras que obligatoriamente deben quedar en un nodo hoja.

Poda (Pruning): Consiste en dejar crecer el árbol completamente y luego recortar las ramas que aportan poco valor predictivo o que se basan en muy pocas muestras.

Módulo 05: Árboles de Regresión
Aunque los árboles se usan comúnmente para clasificación (agrupar en categorías), también pueden resolver problemas de regresión (predecir un número continuo, como el precio de una casa o la temperatura).

Diferencia clave: En lugar de usar la impureza de Gini o la entropía (que miden pureza de clases), los árboles de regresión buscan minimizar la Varianza o el Error Cuadrático Medio (MSE) dentro de cada nodo.

El valor predicho: Cada nodo hoja no devuelve una clase, sino el promedio de los valores numéricos de todas las muestras de entrenamiento que cayeron en esa hoja.

Módulo 06: Limitaciones de los Árboles de Decisión
A pesar de ser modelos de caja blanca muy potentes y fáciles de interpretar, tienen desventajas importantes:

Inestabilidad: Son extremadamente sensibles a pequeños cambios en los datos de entrenamiento. Modificar ligeramente una muestra puede hacer que el árbol cambie por completo su estructura de divisiones.

Ejes ortogonales: Las fronteras de decisión de los árboles siempre son líneas rectas perpendiculares a los ejes coordenados, lo que dificulta modelar relaciones diagonales o circulares complejas de forma eficiente.

Tendencia al sesgo si no se equilibran: Si una clase es mucho más frecuente que otra, el árbol tiende a dominar la estructura a favor de la clase mayoritaria.
