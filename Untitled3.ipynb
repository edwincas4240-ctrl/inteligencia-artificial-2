1. AdaBoost (Adaptive Boosting)
¿Qué es? Es un algoritmo de aprendizaje supervisado perteneciente a la familia de los métodos de ensamblado (ensemble learning). Fue propuesto por Yoav Freund y Robert Schapire en 1995.

¿Qué hace? Combina las predicciones de múltiples clasificadores débiles (generalmente árboles de decisión muy simples de un solo nivel, llamados stumps) para construir un clasificador fuerte y de alta precisión.

Particularidades y funcionamiento:

Secuencial y adaptativo: Entrena modelos uno tras otro. En cada iteración, evalúa los errores del modelo anterior y aumenta el "peso" o la importancia de los datos mal clasificados para que el siguiente modelo se enfoque en ellos.

Votación ponderada: A diferencia del voto simple, asigna mayor peso a las predicciones de aquellos clasificadores individuales que tuvieron menor tasa de error general.

Sensibilidad al ruido: Es muy sensible a datos atípicos (outliers) y datos con ruido, ya que el algoritmo dedicará un esfuerzo excesivo a intentar clasificarlos correctamente en iteraciones posteriores.

2. Retropropagación Recurrente de Almeida–Pineda (Almeida–Pineda Recurrent Backpropagation)
¿Qué es? Es un algoritmo de entrenamiento para redes neuronales recurrentes (RNN) diseñado de forma independiente por Luis B. Almeida y Fernando J. Pineda a finales de la década de 1980.

¿Qué hace? Permite calcular los gradientes del error en redes neuronales con conexiones recurrentes (con retroalimentación) que alcanzan un estado de equilibrio estático antes de calcular la salida.

Particularidades y funcionamiento:

Basado en puntos fijos: Asume que la red actúa como un sistema dinámico disipativo que converge hacia un punto fijo o estado estacionario dado un estímulo de entrada constante.

Eficiencia computacional: A diferencia de BPTT (Backpropagation Through Time), que requiere desplegar la red a lo largo del tiempo aumentando el uso de memoria, Almeida–Pineda calcula el gradiente directamente en el estado de equilibrio mediante la resolución de un sistema de ecuaciones lineales adyacente.

Limitación: Solo es aplicable a problemas donde la dinámica de la red converge a un estado estable, no para secuencias temporales dinámicas cambiantes.

3. ALOPEX (Algorithms of Pattern Extraction)
¿Qué es? Un algoritmo de optimización heurístico y estocástico desarrollado originalmente por Tzanakou et al. (1977) para mapear campos receptivos visuales en neurobiología, adaptado posteriormente para optimizar pesos en redes neuronales.

¿Qué hace? Optimiza los pesos de una red neuronal utilizando únicamente el valor global de una función de costo y perturbaciones aleatorias, sin necesidad de calcular derivadas o gradientes matemáticos.

Particularidades y funcionamiento:

Sin derivadas (Gradient-free): Es ideal para problemas donde la función de costo no es diferenciable o es discontinua.

Uso de correlación temporal: Ajusta un peso según la correlación de sus últimos dos cambios con respecto a los últimos dos cambios en la función de costo global. Si ambos disminuyeron el costo, continúa en esa dirección; de lo contrario, invierte el rumbo.

Control por temperatura estocástica: Incluye un componente de ruido aleatorio regulado por un parámetro similar a la "temperatura" en el recocido simulado (simulated annealing), lo que le permite salir de mínimos locales.

4. Retropropagación (Backpropagation)¿Qué es? El algoritmo fundamental para el entrenamiento de redes neuronales artificiales profundas (feedforward). Popularizado en 1986 por Rumelhart, Hinton y Williams.

¿Qué hace? Calcula el gradiente de la función de pérdida con respecto a cada peso de la red, propagando el error desde la capa de salida hacia atrás hasta la capa de entrada, permitiendo ajustar los pesos mediante descenso de gradiente.

Particularidades y funcionamiento:Regla de la cadena: Aplica de forma iterativa la regla de la cadena del cálculo diferencial para obtener las derivadas parciales de la función de error.

Dos fases por iteración:Forward pass (paso hacia adelante): Calcula la salida proyectada e mide el error.

Backward pass (paso hacia atrás): Propaga el error y actualiza los pesos ajustando la tasa de aprendizaje ($\eta$)

Problemas típicos: Susceptible a quedar atrapado en mínimos locales y propenso a los problemas de gradiente desvanecido (vanishing gradient) o desbordado (exploding gradient) en redes muy profundas.

5. Agregación Bootstrap (Bagging - Bootstrap Aggregating)
¿Qué es? Un meta-algoritmo de ensamblado de aprendizaje automático introducido por Leo Breiman en 1996, diseñado para mejorar la estabilidad y precisión.

¿Qué hace? Reduce la varianza de modelos con tendencia al sobreajuste (overfitting), como los árboles de decisión complejos, combinando las predicciones de varios modelos entrenados en paralelo.

Particularidades y funcionamiento:

Muestreo Bootstrap: Crea múltiples subconjuntos del dataset original realizando un muestreo aleatorio con reemplazo.

Entrenamiento en paralelo: Entrena un modelo independiente en cada subconjunto de datos de forma paralela.

Agregación de resultados: Para predecir, promedia los resultados (en regresión) o realiza un voto de mayoría (en clasificación).

Ejemplo clásico: El algoritmo Random Forest es una extensión de Bagging que además selecciona subconjuntos aleatorios de características.

6. Algoritmo CN2
¿Qué es? Un algoritmo de inducción de reglas desarrollado por Peter Clark y Tim Niblett en 1989 para la clasificación en aprendizaje automático supervisado.

¿Qué hace? Extrae directamente de un conjunto de datos un conjunto de reglas lógicas del tipo "SI condición ENTONCES clase", capaces de clasificar instancias invisibles.

Particularidades y funcionamiento:

Inspiración híbrida: Combina la capacidad de manejo de datos ruidosos de los árboles de decisión (como C4.5) con la interpretabilidad de algoritmos de inducción de reglas anteriores (como AQ).

Búsqueda por haz (Beam Search): Utiliza una búsqueda por haz para explorar eficientemente el espacio de condiciones sin quedar atascado en óptimos locales.

Tolerancia al ruido: Emplea pruebas de significancia estadística (como la entropía o la evaluación Laplacian) para detener la creación de reglas y evitar reglas demasiado específicas que causen sobreajuste.

7. Construcción de Árboles de Habilidades (Skill Tree Construction)
¿Qué es? Un algoritmo o metodología dentro del Aprendizaje por Refuerzo (Reinforcement Learning) y la robótica, utilizado para la abstracción de acciones.

¿Qué hace? Organiza y segmenta trayectorias continuas de habilidades aprendidas por un agente en una estructura jerárquica de árbol de opciones (options).

Particularidades y funcionamiento:

Aislamiento de habilidades: Analiza demostraciones continuas (como movimientos robóticos) y las divide en sub-tareas reutilizables delimitadas por estados clave o metas.

Estructura en árbol: Cada nodo del árbol representa una habilidad específica con su respectiva política de control local y condición de parada.

Transferencia de aprendizaje: Permite que un agente aplique sub-habilidades aprendidas en una tarea previa a nuevos entornos sin necesidad de reentrenar desde cero.

8. Aprendizaje del Árbol de Decisión (Decision Tree Learning)
¿Qué es? Una de las técnicas de modelado predictivo supervisado más populares y no paramétricas, utilizada tanto en clasificación como en regresión.

¿Qué hace? Construye un modelo en forma de diagrama de flujo (árbol) que divide iterativamente los datos según preguntas basadas en las características de entrada hasta llegar a una conclusión o clase.

Particularidades y funcionamiento:

Criterios de división: En cada nodo, busca la variable que mejor divida los datos. Para ello utiliza métricas de impureza como la Entropía / Ganancia de Información (algoritmos ID3, C4.5) o el Índice Gini (algoritmo CART).

Interpretabilidad: Genera modelos altamente explicables por humanos que pueden traducirse fácilmente a reglas lógicas.

Poda (Pruning): Para evitar que el árbol crezca indefinidamente y se sobreajuste a las memorizaciones del dataset, se aplican técnicas de poda previa o posterior.

9. Modelo de Dehaene–Changeux
¿Qué es? Un modelo neurocomputacional y de arquitectura cognitiva propuesto por Stanislas Dehaene y Jean-Pierre Changeux a finales de los años 80 y principios de los 90.

¿Qué hace? Simula circuitos neuronales biológicos en el córtex prefrontal para explicar procesos cognitivos complejos como la flexibilidad ejecutiva, la memoria de trabajo y la toma de decisiones.

Particularidades y funcionamiento:

Inspiración biológica estricta: Modela neuronas que interactúan mediante excitación y la inhibición lateral, emulando la estructura laminar del cerebro.

Mecanismo de Recompensa/Castigo: Incorpora sistemas de aprendizaje por refuerzo modulados por dopamina que ajustan la actividad neuronal en función del éxito o fracaso de una tarea (por ejemplo, en el test de clasificación de cartas de Wisconsin).

Espacio de Trabajo Neuronal Global: Constituyó la base para la teoría del espacio de trabajo global, explicando cómo la información pasa de procesamientos inconscientes a un estado consciente centralizado.

10. Mapa de Difusión (Diffusion Map)¿Qué es? Un algoritmo de reducción de dimensionalidad no lineal y aprendizaje de colectores (manifold learning), introducido por Ronald Coifman y Stephane Lafon en 2006.
¿Qué hace? Proyecta datos complejos de alta dimensión a un espacio de menor dimensión, preservando la geometría subyacente y las relaciones de proximidad locales y globales.
Particularidades y funcionamiento:Proceso de difusión Markoviano: Define una matriz de transición de cadenas de Markov sobre los datos, calculando las probabilidades de "saltar" de un punto a otro basándose en una función de similitud (como el núcleo Gaussiano).
Distancia de Difusión: Mide la distancia entre dos puntos no mediante una línea recta (Euclidiana), sino analizando la cantidad de caminos probables que conectan esos puntos tras transcurrir un tiempo $t$.
Descomposición espectral: Utiliza los autovalores y autovectores (eigenvalues/eigenvectors) de la matriz de transición para construir las nuevas coordenadas reducidas, siendo muy resistente al ruido en los datos.
