Estudio Temático: Entrenando Redes Profundas

Las imágenes muestran una estructura de temario enfocada en los retos y las mejores prácticas para entrenar redes neuronales profundas con éxito. A continuación, se desglosa cada uno de los 6 temas principales:

01. Reto de los Gradientes

Descripción: Al entrenar redes con muchas capas ocultas, surgen problemas críticos con la propagación de los gradientes durante la retropropagación (backpropagation).   

Desglose del tema:

 Desvanecimiento de gradientes (Vanishing Gradients): Ocurre cuando los gradientes se vuelven exponencialmente pequeños a medida que se propagan hacia las capas iniciales, impidiendo que los pesos de esas capas se actualicen de manera significativa y deteniendo el aprendizaje.

 xplosión de gradientes (Exploding Gradients): Se presenta cuando los gradientes crecen demasiado debido a multiplicaciones sucesivas de pesos grandes, provocando inestabilidad numérica o desbordamiento en el modelo.

 Soluciones comunes: Uso de técnicas de recorte de gradientes (gradient clipping), funciones de activación adecuadas y arquitecturas con conexiones residuales (como ResNet).

02. Inicialización

 Descripción: La forma en que se asignan los valores iniciales a los pesos y sesgos de la red antes de comenzar el entrenamiento es crucial para la convergencia.   

 Desglose del tema:

 Problema de la inicialización en ceros: Si todos los pesos se inicializan en cero, todas las neuronas de una capa calcularán la misma salida y gradiente, anulando la capacidad de la red para aprender características diversas.

 Inicialización de Xavier (Glorot): Diseñada para mantener la varianza de las activaciones y los gradientes constante a través de las capas, ideal para funciones de activación simétricas (como tanh o sigmoide).

 Inicialización de He: Variante optimizada específicamente para funciones de activación de tipo ReLU y sus derivadas.

03. Funciones de Activación

 Descripción: Son operaciones matemáticas aplicadas a la salida de cada neurona que introducen no linealidades, permitiendo que la red aprenda patrones complejos.   

 Desglose del tema:

 Sigmoide y Tanh: Clásicas pero propensas al problema del desvanecimiento de gradientes en redes muy profundas.

 ReLU (Rectified Linear Unit): La función más popular en la actualidad debido a su eficiencia computacional y a mitigar el desvanecimiento de gradientes en la región positiva.

 Variantes de ReLU: Opciones como Leaky ReLU, Parametric ReLU (PReLU) o Exponential Linear Unit (ELU) diseñadas para evitar el problema de la "neurona muerta" cuando la entrada es negativa.

04. Normalización

 Descripción: Conjunto de técnicas aplicadas para estabilizar y acelerar el proceso de entrenamiento controlando la distribución de las activaciones internas.   

 Desglose del tema:

 Normalización por lotes (Batch Normalization): Normaliza las entradas de cada capa restando la media del lote y dividiendo por su desviación estándar, lo que permite usar tasas de aprendizaje más altas y reduce la dependencia de una mala inicialización.

 Otras variantes de normalización: Layer Normalization, Instance Normalization y Group Normalization, muy útiles según el tipo de arquitectura (como procesamiento de lenguaje natural o visión por computador).

05. Transferencia (Transfer Learning)

 Descripción: Metodología que aprovecha un modelo preentrenado en un conjunto de datos masivo para resolver una nueva tarea con menor cantidad de datos y recursos computacionales.   

 Desglose del tema:

 Extracción de características: Congelar las capas iniciales y medias de la red preentrenada (que capturan patrones genéricos como bordes o texturas) y entrenar únicamente un nuevo clasificador final para el problema específico.

 Ajuste fino (Fine-Tuning): Descongelar algunas o todas las capas del modelo base y entrenarlas con una tasa de aprendizaje muy baja para adaptar los pesos a las particularidades del nuevo dominio.

06. Preentrenamiento

 Descripción: Estrategia orientada a dotar a la red de representaciones útiles antes de realizar el entrenamiento supervisado enfocado en la tarea final.  

 Desglose del tema:

 Preentrenamiento no supervisado o auto-supervisado: Entrenar el modelo primero en grandes volúmenes de datos sin etiquetar para que aprenda la estructura subyacente de los datos (muy común en modelos de lenguaje modernos y modelos fundacionales).

 Ventajas: Mejora notablemente la capacidad de generalización del modelo y previene el sobreajuste (overfitting) en escenarios con escasez de datos etiquetados.
