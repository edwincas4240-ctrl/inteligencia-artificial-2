Fundamentos Teóricos y Prácticos de Redes Neuronales Artificiales con Keras1. Introducción General y Mapa de Ruta del CursoEl estudio de las redes neuronales artificiales (ANN) constituye uno de los pilares fundamentales del aprendizaje automático moderno y la inteligencia artificial. A través de la estructura del curso expuesta en las diapositivas, el aprendizaje se organiza en una ruta lógica y progresiva:Módulo 01: Inspiración biológica y la transición conceptual de la naturaleza a los modelos computacionales.   Módulo 02: El estudio del Perceptrón, la unidad básica de procesamiento y su evolución histórica desde 1957.   Módulo 03: La expansión hacia el Perceptrón Multicapa (MLP) y la implementación de la retropropagación (backpropagation) para superar límites no lineales.   Módulos 04, 05 y 06: La aplicación práctica en el ecosistema de Keras mediante la API Secuencial, la API Funcional y la API de Subclases.   Módulo 1: Biología a Redes (Inspiración Natural)1.1 La Neurona Biológica y su MorfologíaPara comprender las redes neuronales artificiales, es indispensable observar primero el sistema nervioso biológico, el cual sirve de inspiración arquitectónica. Una neurona biológica consta de tres partes esenciales ilustradas en los diagramas de estudio:Cuerpo celular o Soma: Contiene el núcleo y los orgánulos metabólicos encargados de mantener viva a la célula. Integra todas las señales eléctricas recibidas.Dendritas: Estructuras ramificadas que actúan como las antenas receptoras de la célula. Captan los neurotransmisores químicos liberados por los axones de otras neuronas vecinas.Axón: Una prolongación larga y delgada a través de la cual la neurona transmite su impulso eléctrico de salida hacia otras células mediante las conexiones sinápticas.1.2 Del Sistema Biológico al Modelo ArtificialExiste una equivalencia matemática y funcional directa entre los componentes biológicos y los artificiales:Las dendritas biológicas equivalen a los vectores de entrada ($x_i$) en una neurona artificial.   La fuerza o eficiencia de la conexión sináptica biológica se modela mediante los pesos sinápticos ($w_i$).   El soma recopila las señales y, si el estímulo supera un umbral crítico, se produce el disparo (action potential); esto se replica mediante la función de activación y el sesgo (bias) en la neurona artificial.   Módulo 2: El Perceptrón – La Unidad Fundamental2.1 Historia y Arquitectura (1957)Introducido por Frank Rosenblatt en 1957, el Perceptrón es el ancestro directo de las redes neuronales profundas actuales. Es un algoritmo de clasificación supervisada capaz de separar datos linealmente. Su representación matemática canónica formalizada en las diapositivas es:$$h_{\mathbf{w},b}(\mathbf{x}) = \text{step}(\mathbf{w}^T\mathbf{x} + b)$$Donde:$\mathbf{x}$ representa el vector de entradas.$\mathbf{w}$ representa el vector de pesos sinápticos aplicados a cada entrada.$b$ es el sesgo (bias), que desplaza la función de activación permitiendo mayor flexibilidad al modelo.$\text{step}(z)$ es la función de activación escalón, que devuelve 1 si $z \ge 0$ y 0 en caso contrario.2.2 Componentes de la Unidad Lógica Umbral (TLU)Entradas ($x_1, x_2, x_3$): Los datos numéricos provistos al sistema.   Pesos ($w_1, w_2, w_3$): Coeficientes que determinan la importancia relativa o influencia de cada entrada sobre la decisión final de la neurona.   Función Lineal: Realiza el producto punto o suma ponderada de las entradas y los pesos sumados al sesgo ($z = \sum w_i x_i + b$).   Función Escalón: Transforma la salida continua del cálculo lineal en una decisión discreta (binaria).   2.3 Limitaciones Históricas: El Problema del Computador Frustrado (XOR)Una de las lecciones históricas más importantes expuestas en el curso es la limitación del perceptrón simple para resolver problemas que no son linealmente separables.Cuando se intentan resolver funciones lógicas complejas como el XOR (O exclusivo), los datos forman patrones geométricos que una sola línea recta (o hiperplano) no puede dividir correctamente.Esto sumió a la investigación en redes neuronales en un estancamiento temporal hasta la llegada de métodos multicapa y algoritmos de retropropagación eficientes.Módulo 3: Retropropagación y Perceptrón Multicapa (MLP)3.1 Superando los Límites: El Perceptrón Multicapa (MLP)Para resolver problemas complejos y no lineales (como clasificaciones de imágenes o patrones mixtos), se estructuran múltiples neuronas en capas conectadas secuencialmente:Capa de Entrada: Recibe los datos crudos del conjunto de datos.Capas Ocultas: Capas intermedias que extraen características abstractas y representaciones jerárquicas complejas de los datos.Capa de Salida: Produce la predicción final del modelo (clasificación, regresión, etc.).3.2 El Algoritmo de Retropropagación (Backpropagation)El aprendizaje en un MLP se logra mediante un ciclo iterativo dividido en dos fases fundamentales:Propagación hacia adelante (Forward Pass): Los datos fluyen desde la capa de entrada hasta la de salida, calculando las activaciones de cada neurona hasta obtener una predicción estimada.Propagación hacia atrás (Backward Pass): Se cuantifica el error de la predicción mediante una función de pérdida (loss function). Utilizando cálculo multivariable (específicamente la regla de la cadena de las derivadas), el algoritmo propaga el error de salida hacia atrás a través de la red, midiendo la responsabilidad de cada peso en el error cometido y ajustándolos mediante un optimizador (como el Descenso del Gradiente) para minimizar dicho error en futuras iteraciones.Módulo 4: Arquitecturas de Implementación Práctica en KerasKeras provee un marco de trabajo flexible y de alto nivel sobre TensorFlow para estructurar redes neuronales mediante diferentes enfoques de programación:4.1 API Secuencial (Sequential)Es el enfoque más sencillo y directo. Permite construir modelos apilando capas una tras otra de manera lineal (capa por capa).Uso ideal: Redes feedforward tradicionales donde cada capa tiene exactamente un tensor de entrada y un tensor de salida.4.2 API Funcional (Functional API)Permite crear estructuras de grafos de capas mucho más complejas y flexibles que el modelo secuencial.Uso ideal: Arquitecturas avanzadas que manejan múltiples entradas, múltiples salidas, capas compartidas o conexiones residuales (como las utilizadas en redes ResNet).4.3 API de Subclases (Subclassing API)Otorga un control total mediante la programación orientada a objetos (herencia de la clase tf.keras.Model).Uso ideal: Investigadores y desarrolladores avanzados que necesitan definir bucles de entrenamiento personalizados, comportamientos dinámicos o arquitecturas experimentales complejas ejecutadas de manera imperativa.
