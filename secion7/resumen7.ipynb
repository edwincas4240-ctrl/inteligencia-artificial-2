2. Sección 1: Valor del ML (¿Por qué usar Machine Learning?)
Definición Fundacional
Machine Learning se define como la capacidad de un programa informático de mejorar su rendimiento en una tarea específica a medida que adquiere mayor experiencia.
Ciclo de Minería de Datos y Aprendizaje
El aprendizaje se facilita mediante procesos iterativos de extracción de conocimiento:
1.	Estudio del problema (Study the problem): Identificación del requerimiento o necesidad técnica.
2.	Entrenamiento del modelo (Train ML model): Procesamiento de grandes volúmenes de datos ("Lots of data").
3.	Inspección de la solución (Inspect the solution): Evaluación de resultados y generación de valor.
4.	Comprensión del problema (Better understanding of the problem): Retroalimentación para refinamiento continuo.
5.	Iteración (Repeat if needed): Proceso cíclico hasta obtener el desempeño requerido.
3. Sección 2: Tipos de Sistemas según el Nivel de Supervisión
Los sistemas de ML se clasifican según cómo se procesan y etiquetan los datos:
Tipo de Aprendizaje	Características y Flujo de Trabajo	Representación Visual
Supervisado	El modelo entrena con datos previamente etiquetados o clasificados (pares entrada-salida).	Inspección y clasificación directa mediante reglas/etiquetas explícitas.
No Supervisado	El sistema procesa datos sin etiquetar para descubrir patrones, agrupamientos o estructuras subyacentes.	Filtrado y extracción de patrones ocultos desde un flujo constante de datos.
Por Refuerzo	Un agente interactúa con un entorno y aprende una estrategia óptima mediante un sistema de recompensas o penalizaciones.	Ciclo: Observar estado $\rightarrow$ Ejecutar acción $\rightarrow$ Recibir recompensa/penalización $\rightarrow$ Actualizar política.
4. Sección 3: Flujos de Datos y Ritmos de Aprendizaje
Esta sección aborda cómo los datos alimentan a los modelos y cómo varía la frecuencia de actualización según el diseño del sistema:
•	Batch Learning (Aprendizaje por Lotes): Procesamiento masivo de bloques de datos acumulados (Data Block). El modelo se entrena periódicamente de forma fuera de línea (offline).
•	Online/Stream Learning (Aprendizaje Continuo): El sistema recibe y procesa flujos continuos de datos en tiempo real, adaptando de forma incremental sus parámetros y ritmo de aprendizaje.
5. Próximos Pasos en el Plano de ML
Las diapositivas restantes de la presentación cubren las fases finales del ciclo de vida de un sistema de ML:
•	04. Estrategias Generales: Selección de algoritmos, diseño de arquitectura e ingeniería de características.
•	05. Desafíos Principales: Manejo del sobreajuste (overfitting), subajuste (underfitting), sesgos en los datos y la degrada del modelo con el tiempo (data drift).
•	06. Pruebas y Validación: Métricas de rendimiento, validación cruzada y monitoreo continuo en producción.
