CRISP-DM: El Plano Maestro
CRISP-DM es el estándar metodológico interindustrial más utilizado para guiarse en el desarrollo de proyectos de minería de datos, ciencia de datos y aprendizaje automático. Proporciona una hoja de ruta estructurada e iterativa que transforma los datos sin procesar en conocimiento aplicado y valor de negocio.
1. Orígenes y Evolución Histórica
La evolución de la metodología abarca tres hitos clave (Imágenes 12 y 13):
•	1996 – Concepción: Nace a partir del trabajo conjunto de empresas líderes como DaimlerChrysler, SPSS y NCR.
•	1999 – Publicación Oficial: Se lanza formalmente la versión CRISP-DM 1.0, consolidándose rápidamente como el estándar interindustrial (standard inter-industrias) preferido a nivel mundial.
•	2015 – Extensión por IBM: IBM introduce la extensión ASUM-DM (Analytics Solutions Unified Method for Data Mining), refinando la metodología para integrar prácticas avanzadas de despliegue y gestión de proyectos analíticos.
2. Estructura de la Metodología (Agenda General)
La presentación divide la metodología en cinco grandes bloques de trabajo (Imagen 14):
1.	Orígenes y Práctica: Antecedentes, contexto histórico y propósito del marco de trabajo.
2.	Negocio y Datos:
o	Comprensión del Negocio (Business Understanding): Definición de objetivos, requisitos y criterios de éxito del proyecto.
o	Comprensión de los Datos (Data Understanding): Recolección inicial, exploración y verificación de la calidad de las fuentes de datos.
3.	Preparación de Datos (Data Preparation): Selección, limpieza, transformación, construcción de características (feature engineering) y estructuración del conjunto de datos final para el modelo.
4.	Modelado y Evaluación (Modeling & Evaluation):
o	Modelado: Selección de algoritmos, ajuste de hiperparámetros y entrenamiento.
o	Evaluación: Verificación de que los resultados técnicos cumplan con los objetivos de negocio planteados en la fase 1.
5.	Despliegue y Futuro (Deployment): Integración de la solución en producción, monitoreo continuo, mantenimiento y planificación de iteraciones futuras.
Fase 3: Preparación de Datos (Data Preparation)
Esta fase suele consumir entre el 70% y el 80% del tiempo total de un proyecto de ciencia de datos. Su objetivo es tomar los datos en bruto y transformarlos en el conjunto de datos final (final dataset) listo para alimentar los algoritmos de modelado.
Pasos Detallados:
1.	Selección de Datos (Data Selection):
o	Decidir qué tablas, atributos y registros se van a utilizar y cuáles se descartan.
o	Criterios de inclusión/exclusión basados en la calidad, relevancia técnica y restricciones legales o de privacidad.
2.	Limpieza de Datos (Data Cleaning):
o	Tratamiento de valores faltantes (Missing Values): Imputación (por media, mediana, moda o modelos predictivos) o eliminación de registros dañados.
o	Manejo de atípicos (Outliers): Detección mediante métodos estadísticos (z-score, IQR) y decisión sobre corregirlos, aislarlos o eliminarlos.
o	Corrección de inconsistencias: Normalización de formatos (fechas, texto en mayúsculas/minúsculas, tipos de datos).
3.	Construcción de Datos (Feature Engineering):
o	Creación de nuevas variables derivadas que ayuden al modelo a aprender mejor (ej. extraer el día de la semana a partir de una fecha, o calcular la razón de endeudamiento a partir de ingresos y deudas).
o	Transformación de variables categóricas a numéricas mediante One-Hot Encoding o Label Encoding.
4.	Integración de Datos (Data Integration):
o	Combinación de información proveniente de múltiples tablas o fuentes heterogéneas mediante operaciones de cruce (Joins, Merges) y agregación.
5.	Formateo y Escalado de Datos (Data Formatting & Scaling):
o	Normalización o estandarización de variables numéricas (como MinMax Scaler o StandardScaler) para evitar que atributos con valores grandes dominen sobre los pequeños.
Fase 5: Despliegue y Futuro (Deployment & Future)
Haber creado un modelo preciso no sirve de nada si el negocio no puede usarlo para tomar decisiones o automatizar tareas. Esta fase lleva el modelo desde un entorno de prueba a la ejecución en el mundo real.
Pasos Detallados:
1.	Planificación del Despliegue (Deployment Plan):
o	Definir la arquitectura de integración: ¿Será una API REST en la nube, un proceso por lotes (batch) nocturno, o un modelo edge dentro de un dispositivo móvil?
o	Establecer estrategias de despliegue como Canary Deployment (desplegar a un % pequeño de usuarios) o Shadow Deployment (ejecutar en paralelo con el sistema anterior para comparar resultados).
2.	Monitoreo y Mantenimiento (Monitoring & Maintenance):
o	Seguimiento de métricas de rendimiento: Evaluación continua de la precisión en producción.
o	Detección de Deriva (Data & Concept Drift): Identificar cuando los datos de entrada reales cambian respecto a los datos con los que el modelo fue entrenado (ej. cambios en comportamiento de consumidores tras una crisis).
3.	InFORME y Entrega de Resultados (Final Report):
o	Documentación técnica del modelo para los equipos de ingeniería/TI.
o	Presentación de resultados y métricas de impacto operativo para los líderes de negocio.
4.	Revisión del Proyecto y Próximos Pasos (Project Review):
o	Evaluar qué funcionó bien y qué falló durante el ciclo de CRISP-DM.
o	Planificar iteraciones futuras, reentrenamientos periódicos o nuevas funcionalidades que requieran iniciar nuevamente el ciclo desde la fase de negocio.
